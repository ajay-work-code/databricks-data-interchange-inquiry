```sql
-- run ONCE, before 01_bronze_setup.sql

DECLARE OR REPLACE VARIABLE catalog_name  STRING = 'bis_dev';
DECLARE OR REPLACE VARIABLE bronze_schema STRING = 'bronze_roster';
DECLARE OR REPLACE VARIABLE silver_schema STRING = 'silver_roster';
DECLARE OR REPLACE VARIABLE volume_name   STRING = 'manual_inputs';

EXECUTE IMMEDIATE 'CREATE SCHEMA IF NOT EXISTS '
    || catalog_name || '.' || bronze_schema;

EXECUTE IMMEDIATE 'CREATE SCHEMA IF NOT EXISTS '
    || catalog_name || '.' || silver_schema;

EXECUTE IMMEDIATE 'CREATE VOLUME IF NOT EXISTS '
    || catalog_name || '.' || bronze_schema || '.' || volume_name;



DROP TABLE bis_dev.bronze_roster.autoloader_audit_log;
DROP TABLE bis_dev.bronze_roster.emp_storage_location;
DROP TABLE bis_dev.bronze_roster.roster_qa;
DROP MATERIALIZED VIEW IF EXISTS bis_dev.silver_roster.emp_storage_location;
DROP VOLUME IF EXISTS bis_dev.bronze_roster.manual_inputs;


create schema if not exists bis_dev.bronze_roster;
create schema if not exists bis_dev.silver_roster;
create volume if not exists bis_dev.bronze_roster.manual_inputs;
```
