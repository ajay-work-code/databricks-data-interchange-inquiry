```sql
DECLARE catalog_name STRING DEFAULT 'bis_dev';
DECLARE bronze_schema STRING DEFAULT 'bronze_roster';

-- Create schema if it doesn't exist
EXECUTE IMMEDIATE
  'CREATE SCHEMA IF NOT EXISTS ' ||
  catalog_name || '.' || bronze_schema;

-- Create table
EXECUTE IMMEDIATE
  'CREATE OR REPLACE TABLE ' || catalog_name || '.' || bronze_schema || '.emp_storage_location (

    -- Source columns
    Rep_Name             STRING    COMMENT ''Name of representative as provided by storage vendor'',
    Rep_Email            STRING    COMMENT ''Incyte Email of representative as provided by storage vendor'',
    Rep_Phone            STRING    COMMENT ''Phone of representative as provided by storage vendor'',
    Territory_ID         STRING    COMMENT ''Territory Number of representative as provided by storage vendor'',
    Emp_ID               STRING    COMMENT ''Employee Number of representative as provided by storage vendor'',
    Facility_Name        STRING    COMMENT ''Storage Location Name'',
    Facility_Raw_Address STRING    COMMENT ''Storage Location Address'',
    Space_Number         STRING    COMMENT ''Storage Location Number'',
    Storage_Size         STRING    COMMENT ''Storage Location Size'',
    Territory_Name       STRING    COMMENT ''Territory Name of representative as provided by storage vendor'',
    Region_Emp_Name      STRING    COMMENT ''ASD or OBD name as provided by storage vendor'',
    Region_Name          STRING    COMMENT ''Region Name of representative as provided by storage vendor'',

    -- Parsed address columns
    Address_1            STRING    COMMENT ''Street address parsed from Facility_Raw_Address'',
    Address_2            STRING    COMMENT ''Suite type and number parsed from Facility_Raw_Address, null when no suite is present'',
    City                 STRING    COMMENT ''City parsed from Facility_Raw_Address'',
    State                STRING    COMMENT ''State parsed from Facility_Raw_Address'',
    Zip_Code             STRING    COMMENT ''ZIP code parsed from Facility_Raw_Address'',
    Parsing_Error        STRING    COMMENT ''Address parse failure reason, null when parsing succeeded. Null does not guarantee a correct parse'',

    -- Load lineage
    File_Date            DATE      COMMENT ''Date the file was loaded, taken from the source file modification time on the volume'',
    File_Name            STRING    COMMENT ''Name of the file loaded'',
    Load_Timestamp       TIMESTAMP COMMENT ''Time the ingest job wrote the row. Used to order rows within a File_Date'',
    _rescued_data        STRING    COMMENT ''Rescue data column. Unexpected columns found in the source file, as JSON. Detection is column-level: a new, renamed or extra header''

  )
  COMMENT ''Weekly storage locations supplied by CareTria. Raw vendor columns plus parsed address components. Append only.''';
```
